# To Check Multicollinearity

```python
import statsmodels.api as sm
# Function to calculate VIF
def calculate_vif(data):
    features = data.drop(columns=['Life_expectancy','Year','Country','Region','Economic Status'])
    vif_data = pd.DataFrame()
    vif_data["Feature"] = features.columns
    vif_data["VIF"] = [sm.OLS(features[col], sm.add_constant(features.drop(columns=[col]))).fit().rsquared for col in features]
    return vif_data

# Calculating VIF
vif = calculate_vif(data)
print(vif)
```

# Baseline Model

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Calculating the mean or median of the target variable
baseline_prediction = np.mean(y_train)

# Creating an array of baseline predictions with the same length as the test set
baseline_predictions = np.full_like(y_test, fill_value=baseline_prediction)

# Calculating Mean Squared Error (MSE) for the baseline model
baseline_mse = np.mean((y_test - baseline_predictions) ** 2)

# Compute MAE
mae = mean_absolute_error(y_test, baseline_predictions)

# Compute MAPE
mape = np.mean(np.abs((y_test - baseline_predictions) / y_test)) * 100

print("MAE:", mae)
print("MAPE:", mape)
print("Baseline Mean Squared Error:", baseline_mse)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Obtaining predictions and residuals
X_test_with_const = sm.add_constant(X_test)

# Making predictions using the fitted model
predictions = model.predict(X_test_with_const)
residuals = y_test - predictions

sns.residplot(x=predictions, y=residuals, lowess=True, line_kws={'color': 'red', 'lw': 1})
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

# Normality Check of Residuals 

```python
import scipy.stats as stats
stats.shapiro(residuals)

plt.hist(residuals, bins=20, color='skyblue', edgecolor='black')

plt.xlabel('Residuals')
plt.ylabel('Frequency')
plt.title('Histogram of Residuals')

plt.show()
```

```python
from scipy.stats import shapiro

# Performing Shapiro-Wilk test for normality
statistic, p_value = shapiro(residuals)

# Printing the test statistic and p-value
print("Shapiro-Wilk Test Statistic:", statistic)
print("P-value:", p_value)

# Checking if the p-value is less than the significance level (e.g., 0.05) to determine significance
if p_value < 0.05:
    print("The residuals are not normally distributed (reject null hypothesis)")
else:
    print("The residuals are normally distributed (fail to reject null hypothesis)")
```

# Checking Homoscedasticity of Residuals

```python
from statsmodels.stats.diagnostic import het_breuschpagan

import statsmodels.api as sm

# Adding a constant column to X_test
X_test_with_const = sm.add_constant(X_test)

lm_statistic, lm_p_value, f_statistic, f_p_value = het_breuschpagan(residuals, X_test_with_const)

#Print the results
print("Breusch-Pagan Test Results:")
print("LM Statistic:", lm_statistic)
print("LM P-value:", lm_p_value)
print("F Statistic:", f_statistic)
print("F P-value:", f_p_value)
```

# Cross Validation

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


model = LinearRegression()
model.fit(X_train, y_train)

# Performing k-fold cross-validation
k = 10  # Number of folds
cv_scores = cross_val_score(model, x, y, cv=k)
cv_scores

# Plotting the cross-validation scores
plt.figure(figsize=(8, 6))
plt.plot(np.arange(1, k+1), cv_scores, marker='o', linestyle='-')
plt.xlabel('Fold')
plt.ylabel('Cross-Validation Score')
plt.title('Cross-Validation Scores for {}-fold Cross-Validation'.format(k))
plt.xticks(np.arange(1, k+1))
plt.grid(True)
plt.show()

# Calculating additional metrics
cv_r2_scores = cross_val_score(model, x, y, cv=k, scoring='r2')
cv_rmse_scores = np.sqrt(-cross_val_score(model, x, y, cv=k, scoring='neg_mean_squared_error'))
cv_mse_scores = -cross_val_score(model, x, y, cv=k, scoring='neg_mean_squared_error')

# Create a DataFrame to store the scores
data = {
    'Fold': np.arange(1, k+1),
    'CV Scores': cv_scores,
    'CV R2 Scores': cv_r2_scores,
    'CV RMSE Scores': cv_rmse_scores,
    'CV MSE Scores': cv_mse_scores
}
df_scores = pd.DataFrame(data)

# Calculate mean scores
mean_scores = df_scores.mean().to_dict()
mean_scores['Fold'] = 'Mean'
df_mean_scores = pd.DataFrame([mean_scores])

# Concatenate mean scores to the DataFrame
df_scores = pd.concat([df_scores, df_mean_scores])

# Print the scores in table format
print(df_scores.to_string(index=False))

# Calculating the mean R2 score across all folds
mean_r2_score = np.mean(cv_r2_scores)

# Finding the index of the fold with the highest mean R2 score
best_fold_index = np.argmax(cv_r2_scores)

# Printing the best fold index and its mean R2 score
print("Best Fold Index:", best_fold_index)
print("Mean R2 Score for Best Fold:", cv_r2_scores[best_fold_index])

# Plotting the cross-validation scores
plt.figure(figsize=(8, 6))
plt.plot(np.arange(1, k+1), cv_r2_scores, marker='o', linestyle='-', label='R2 Score')
plt.plot(np.arange(1, k+1), cv_rmse_scores, marker='o', linestyle='-', label='RMSE')
plt.plot(np.arange(1, k+1), cv_mse_scores, marker='o', linestyle='-', label='MSE')
plt.xlabel('Fold')
plt.ylabel('Score')
plt.title('Cross-Validation Scores for {}-fold Cross-Validation'.format(k))
plt.xticks(np.arange(1, k+1))
plt.legend()
plt.grid(True)
plt.show()

# Making predictions on the training set
y_pred_train = model.predict(X_train)

# Calculating evaluation metrics for the training set
mse_train_CV = mean_squared_error(y_train, y_pred_train)
mae_train_CV = mean_absolute_error(y_train, y_pred_train)
r2_train_CV = r2_score(y_train, y_pred_train)
rmse_train_CV = np.sqrt(mse_train_CV)

# Printing evaluation metrics for the training set
print("\nTraining Set:")
print("MSE_CV:", mse_train_CV)
print("MAE_CV:", mae_train_CV)
print("R2_Score_CV:", r2_train_CV)
print("RMSE_CV:", rmse_train_CV)

# Making predictions on the test set
y_pred_test = model.predict(X_test)

# Calculating evaluation metrics for the test set
mse_test_CV = mean_squared_error(y_test, y_pred_test)
mae_test_CV = mean_absolute_error(y_test, y_pred_test)
r2_test_CV = r2_score(y_test, y_pred_test)
rmse_test_CV = np.sqrt(mse_test_CV)

# Printing evaluation metrics for the test set
print("\nTest Set:")
print("MSE_CV:", mse_test_CV)
print("MAE_CV:", mae_test_CV)
print("R2_Score_CV:", r2_test_CV)
print("RMSE_CV:", rmse_test_CV)

# Making predictions on the validation set (assuming X_val and y_val are defined)
y_pred_val = model.predict(X_val)

# Calculating evaluation metrics for the validation set
mse_val_CV = mean_squared_error(y_val, y_pred_val)
mae_val_CV = mean_absolute_error(y_val, y_pred_val)
r2_val_CV = r2_score(y_val, y_pred_val)
rmse_val_CV = np.sqrt(mse_val_CV)

# Printing evaluation metrics for the validation set
print("\nValidation Set:")
print("MSE_CV:", mse_val_CV)
print("MAE_CV:", mae_val_CV)
print("R2_Score_CV:", r2_val_CV)
print("RMSE_CV:", rmse_val_CV)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import mean_squared_error

# Defining the parameter grid including n_estimators
param_grid = {
    'n_estimators': [100]  # Specify different values for n_estimators
}

# Initializing the Gradient Boosting Regressor
gb_reg = GradientBoostingRegressor(random_state=42)

# Initializing Grid Search CV
grid_search = GridSearchCV(gb_reg, param_grid, cv=5, scoring='neg_mean_squared_error')

# Performing Grid Search CV
grid_search.fit(X_train, y_train)

# Geting the best hyperparameters
best_params = grid_search.best_params_

# Training the final model using the best hyperparameters
final_model = GradientBoostingRegressor(**best_params, random_state=42)
final_model.fit(X_train, y_train)

# Making predictions
y_train_pred = final_model.predict(X_train)
y_test_pred = final_model.predict(X_test)
y_val_pred =  final_model.predict(X_val)

# Calculating evaluation metrics for train set
train_mse_GB = mean_squared_error(y_train, y_train_pred)
train_mae_GB = mean_absolute_error(y_train, y_train_pred)
train_r2_GB = r2_score(y_train, y_train_pred)
train_rmse_GB = np.sqrt(train_mse_GB)

# Calculating evaluation metrics for test set
test_mse_GB = mean_squared_error(y_test, y_test_pred)
test_mae_GB = mean_absolute_error(y_test, y_test_pred)
test_r2_GB = r2_score(y_test, y_test_pred)
test_rmse_GB = np.sqrt(test_mse_GB)

# Calculating evaluation metrics for validation set
val_mse_GB = mean_squared_error(y_val, y_val_pred)
val_mae_GB = mean_absolute_error(y_val, y_val_pred)
val_r2_GB = r2_score(y_val, y_val_pred)
val_rmse_GB = np.sqrt(val_mse_GB)

print("Train set:")
print("MSE_GB:", train_mse_GB)
print("MAE_GB:", train_mae_GB)
print("R2 Score_GB:", train_r2_GB)
print("RMSE_GB:", train_rmse_GB)

print("\nTest set:")
print("MSE_GB:", test_mse_GB)
print("MAE_GB:", test_mae_GB)
print("R2 Score_GB:", test_r2_GB)
print("RMSE_GB:", test_rmse_GB)

print("\nValidation set:")
print("MSE_GB:", test_mse_GB)
print("MAE_GB:", test_mae_GB)
print("R2 Score_GB:", test_r2_GB)
print("RMSE_GB:", test_rmse_GB)
```
